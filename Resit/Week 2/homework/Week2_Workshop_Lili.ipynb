### Exercise 1: Clean Water Countdown
⏱ **15 minutes** | Slide 11

**Context:** You will simulate a water purifier with numbered cycles (SDG 6: Clean Water). Printed messages only describe this classroom simulation. They do not establish that real water is safe to drink.

#### Step 1: Run the cycles
Use a `while` loop to run **8 purification cycles**. Print `'Cycle X: water batch purified'` for each cycle, and a final message when all cycles are done.

> ⚠️ If your loop never stops: press the ⏹ (interrupt) button. Something inside the loop must change the condition!

**Slide reminder (5–10):** A `while` loop checks before each repetition. Trace the starting value, the check, the indented block, and the update. The line after the block runs when the loop ends.


**My prediction:** it prints `cycle 1`, `cycle 2`, `cycle 3` and then stops, because after the third round `n` becomes 4 and `n <= 3` is false.

````python
# Predict-run-check
n = 1
while n <= 3:
    print('cycle', n)
    n = n + 1
````

````python
# Step 1: Run 8 purification cycles with a while loop
cycle = 1
while cycle <= 8:
    print('Cycle ' + str(cycle) + ': water batch purified')
    cycle = cycle + 1

print('All 8 cycles done (classroom simulation only).')
````

#### Step 2: Validate the input
The purifier needs purification tablets. Ask the operator how many tablets are left. Keep asking until you get a number of **1 or more** — use the standard pattern `while True:` + `break`.

**Slide 9 reminder:** `input()` returns text. `int()` converts whole-number text to a number. For this exercise, enter integer text and test whether the number is allowed. Non-number text raises `ValueError`; handling that is outside this basic validation task.


````python
# Step 2: Input validation with while True + break
while True:
    tablets = int(input('How many purification tablets are left? '))
    if tablets >= 1:
        break
    print("That can't be right - enter a number of 1 or more.")

print('Thanks! Tablets in stock:', tablets)
````

#### Step 3: Skip a cycle
Every 5th cycle the filter must be cleaned, so **no water is purified in cycle 5**. Copy your loop from Step 1 and use `continue` to skip the purification message in cycle 5 (print `'Cycle 5: filter cleaning — no water purified.'` instead).

**Slide 10 reminder:** `break` leaves the loop. `continue` skips the rest of this repetition. Make sure the counter can still change before the next check.


````python
# Step 3: Skip cycle 5 with continue
cycle = 0
while cycle < 8:
    cycle = cycle + 1          # increase first, so continue can't skip it
    if cycle == 5:
        print('Cycle 5: filter cleaning - no water purified.')
        continue
    print('Cycle ' + str(cycle) + ': water batch purified')

print('All cycles done.')
````

### Exercise 2: Vaccination Campaign Tracker
⏱ **15 minutes** | Slide 16

**Context:** A clinic runs a vaccination campaign and can vaccinate **250 people per day** (SDG 3: Good Health & Well-being). A `for` loop plus an *accumulator variable* is the basic shape of all data processing.

#### Part A: Track the campaign
1. Create `total_vaccinated = 0`.
2. Write a `for` loop over `range(1, 8)` (days 1–7).
3. Each day, add 250 to the total and print: `'Day X - total vaccinated: Y'`.

**Slides 12–15 reminder:** `range()` stops before its stop value. An accumulator remembers a running total. Initialise it before the loop, then update it inside the loop.


````python
# Part A: 7-day campaign with an accumulator
total_vaccinated = 0
for day in range(1, 8):
    total_vaccinated = total_vaccinated + 250
    print('Day', day, '- total vaccinated:', total_vaccinated)
````

#### Part B: Reach the target
1. Ask the user for a campaign target with `input()` (e.g. 10000 people).
2. Use a `while` loop to count how many days are needed at 250 vaccinations/day.
3. Print the result.
4. **Bonus:** every 2nd day (day 2, 4, 6) a mobile clinic joins. Print those days using `range(start, stop, step)`.

````python
# Part B: How many days to reach the target?
target = int(input('Campaign target (people): '))

vaccinated = 0
days = 0
while vaccinated < target:
    days = days + 1
    vaccinated = vaccinated + 250

print('Days needed to reach', target, 'people:', days)

# Bonus: mobile clinic days with range(start, stop, step)
for day in range(2, 8, 2):
    print('Day', day, '- the mobile clinic joins')
````

## Learning Wave 3 — Functions & Scope

### Exercise 3: Health Risk Screener
⏱ **20 minutes** | Slide 22

**Context:** Hospitals and insurers use **BMI** (Body Mass Index) to screen health risk: `bmi = weight / height²`. The WHO categories are:

| BMI | Category |
|---|---|
| < 18.5 | underweight |
| 18.5 – 24.9 | healthy weight |
| 25 – 29.9 | overweight |
| ≥ 30 | obese |

**Steps:**
1. Write a function `calculate_bmi(weight_kg, height_m)` that **returns** the BMI. (Power operator: `height_m ** 2`.)
2. Write a function `classify_bmi(bmi)` that returns the WHO category using `if/elif/else`.
3. Ask the user for weight and height, call both functions, and print the result.
4. **Scope check:** after the cell runs, try printing a variable that only exists *inside* a function. What happens?

> 💡 `round(x, 1)` rounds a number to 1 decimal — handy for printing.

**Slides 17–21 reminder:** `def` defines the function; a call runs it. Parameters receive inputs. `return` sends a value back, while `print()` displays it. For the scope check, choose a name that is not already defined outside the function. Reserve the final **5 minutes** for the reflection below.


````python
# Exercise 3: Health Risk Screener

# Step 1: define calculate_bmi(weight_kg, height_m)
def calculate_bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

# Step 2: define classify_bmi(bmi) with if/elif/else
def classify_bmi(bmi):
    if bmi < 18.5:
        return 'underweight'
    elif bmi < 25:
        return 'healthy weight'
    elif bmi < 30:
        return 'overweight'
    else:
        return 'obese'

# Step 3: ask for weight + height, call both functions, print the result
weight = float(input('Weight in kg: '))
height = float(input('Height in m: '))
bmi = calculate_bmi(weight, height)
print('BMI:', round(bmi, 1), '-', classify_bmi(bmi))

# Step 4: scope check - weight_kg only exists inside calculate_bmi
try:
    print(weight_kg)
except NameError as error:
    print('Scope check:', error)
    print('weight_kg is a local variable: it only exists while calculate_bmi runs.')
````

#### ✍️ Reflection

### 🧠 Mood Tracker — A Mental Health Check-in Bot

**Goal:** Build an interactive program that checks in on someone's mood every day for a week, shows the week as a chart, and gives personalised feedback.

**Context:** Use fictional mood scores to practise repeated input, functions and a weekly summary. This classroom program does not diagnose a condition. Consider what a single score leaves out.

#### Requirements:

| # | Requirement | Concepts |
|---|---|---|
| 1 | A function `get_valid_score(day)` that asks for a mood score and keeps asking until it gets a number from 1–10, then **returns** it | `while True` + `break`/`return`, validation |
| 2 | A `for` loop that collects scores for **7 days** and adds them to a total | `for`, `range()`, accumulator |
| 3 | Inside the loop, print a bar for each day: `Day 3 : ####### (7)` — use string multiplication `'#' * score` | loops + Week 1 strings |
| 4 | Track the **best** and **hardest** day using comparison variables (no lists — that's next week!) | `if`, variables |
| 5 | A function `calculate_average(total, days)` that returns the average | `def`, `return` |
| 6 | A function `give_feedback(average)` that returns a personalised message using `if/elif/else` | functions + conditionals |
| 7 | Print a final summary: average, best day, hardest day, feedback | `print()`, `round()` |

#### Expected output example:
```
Day 1 - how was your mood (1-10)? 6
Day 1 : ###### (6)
Day 2 - how was your mood (1-10)? 11
Please enter a number from 1 to 10.
Day 2 - how was your mood (1-10)? 4
Day 2 : #### (4)
...

--- Your Week in Review ---
Average mood:  5.9
Best day:      day 6 (8/10)
Hardest day:   day 2 (4/10)
A steady week. Small things (sleep, a walk, a friend) can nudge it up.
```

#### Bonus:
- **Streak detector:** count the longest streak of consecutive "good days" (score ≥ 6) — two extra variables in the loop are enough.

#### Required reflection
Use the markdown cell below to discuss whether a bot should give mental health advice. We revisit this together at 16:40 (slide 30).

#### Tips:
1. Build it in pieces: first the input function, then the loop, then the summary.
2. Test your validation with bad input on purpose (0, 11, 999).
3. Stuck in an infinite loop? Interrupt the kernel (⏹) and check what changes inside the loop.
4. Your `if/elif/else` in `give_feedback` must cover **all** possible averages.

**Slides 26–28:** Keep the seven requirements above. Test one part before joining the parts. Try invalid scores `0`, `11`, `999` and valid endpoints `1`, `10`. Check exactly seven daily outputs, verify your total with fictional data, and explain your choice when best or hardest days tie. The streak detector remains optional.


````python
# Big Assignment: Mood Tracker - A Mental Health Check-in Bot

# --- Step 1: define get_valid_score(day) ---
def get_valid_score(day):
    while True:
        score = int(input('Day ' + str(day) + ' - how was your mood (1-10)? '))
        if 1 <= score <= 10:
            return score
        print('Please enter a number from 1 to 10.')

# --- Step 2: define calculate_average(total, days) ---
def calculate_average(total, days):
    return total / days

# --- Step 3: define give_feedback(average) ---
def give_feedback(average):
    if average >= 8:
        return 'A great week! Notice what helped, so you can repeat it.'
    elif average >= 6:
        return 'A steady week. Small things (sleep, a walk, a friend) can nudge it up.'
    elif average >= 4:
        return 'A mixed week. Be kind to yourself and plan one thing you enjoy.'
    else:
        return 'A hard week. Please talk to someone you trust, or to your student counsellor.'

# --- Step 4: the main loop (7 days) ---
total = 0
best_day = 0
best_score = 0
hardest_day = 0
hardest_score = 11
streak = 0          # bonus: current run of good days (score >= 6)
longest_streak = 0

for day in range(1, 8):
    score = get_valid_score(day)
    print('Day', day, ':', '#' * score, '(' + str(score) + ')')
    total = total + score

    # ties: the first day with that score is kept (> and <, not >= and <=)
    if score > best_score:
        best_score = score
        best_day = day
    if score < hardest_score:
        hardest_score = score
        hardest_day = day

    if score >= 6:
        streak = streak + 1
        if streak > longest_streak:
            longest_streak = streak
    else:
        streak = 0

# --- Step 5: print the summary ---
average = calculate_average(total, 7)
print()
print('--- Your Week in Review ---')
print('Average mood: ', round(average, 1))
print('Best day:      day', best_day, '(' + str(best_score) + '/10)')
print('Hardest day:   day', hardest_day, '(' + str(hardest_score) + '/10)')
print('Longest streak of good days (6+):', longest_streak)
print(give_feedback(average))
````

````python
# Testing the validation with the values from the tips:
# 999 must be rejected, and the edges 1 and 10 must be accepted.
print("Returned:", get_valid_score(1))
print("Returned:", get_valid_score(2))
````

## 5 Things to Remember

1. **A `while` loop checks its condition before every repetition.** Make progress toward stopping, or use an explicit exit. Interrupt a loop that keeps running.
2. **A `for` loop takes the next value in turn.** `range(start, stop, step)` stops before `stop`.
3. **`break` leaves the loop. `continue` skips the rest of the current repetition.**
4. **A function runs when called.** Parameters receive inputs, and `return` gives a result to the caller. `print()` displays a value. Without `return`, a function returns `None`.
5. **A working program can still apply a harmful rule.** Our rules are written by people, not learned by a trained model. Ask what context is missing and who is responsible.
